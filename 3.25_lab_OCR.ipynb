```python
# Task 3 – Structured OCR with Functions

# Install dependencies (only needed once per session)
!apt-get install -y tesseract-ocr
!pip install pytesseract pillow

import pytesseract
from PIL import Image
from google.colab import files

def upload_image():
    """Allow user to upload an image file."""
    print("Please upload an image file (e.g., receipt, sign, document):")
    uploaded = files.upload()
    return next(iter(uploaded))

def extract_text_from_image(image_path):
    """Open the image and extract text using Tesseract OCR."""
    img = Image.open(image_path)
    text = pytesseract.image_to_string(img)
    return text.strip()

def display_output(text):
    """Print the extracted text with a clear header."""
    print("\n" + "="*50)
    print("EXTRACTED TEXT")
    print("="*50)
    if text:
        print(text)
    else:
        print("[No text detected]")

def main():
    """Main workflow: upload → extract → display."""
    filename = upload_image()
    print(f"\nProcessing: {filename}")

    try:
        extracted = extract_text_from_image(filename)
        display_output(extracted)
    except Exception as e:
        print(f"\nError: {e}")

if __name__ == "__main__":
    main()
```

```python
# Task 5 – Improved OCR with Preprocessing & Text Cleaning

!apt-get install -y tesseract-ocr
!pip install pytesseract pillow opencv-python

import pytesseract
import cv2
import numpy as np
from PIL import Image
from google.colab import files
import re

def preprocess_image(image_path):
    """Load image, convert to grayscale, apply thresholding and denoising."""
    # Read with OpenCV
    img = cv2.imread(image_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    # Apply Gaussian blur to reduce noise
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    # Thresholding (Otsu)
    _, thresh = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    # Morphological operations to remove small noise
    kernel = np.ones((1, 1), np.uint8)
    cleaned = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)
    return Image.fromarray(cleaned)

def extract_text(image):
    """Extract text using Tesseract."""
    return pytesseract.image_to_string(image)

def clean_text(text):
    """Remove excessive whitespace and non‑alphanumeric noise."""
    # Remove multiple newlines
    text = re.sub(r'\n\s*\n', '\n', text)
    # Replace multiple spaces with single space
    text = re.sub(r' +', ' ', text)
    # Remove lines that are only symbols or very short noise
    lines = [line.strip() for line in text.splitlines() if len(line.strip()) > 2 and not re.match(r'^[\W_]+$', line.strip())]
    return '\n'.join(lines)

def main():
    print("Upload a difficult image (blurry, handwritten, angled):")
    uploaded = files.upload()
    filename = next(iter(uploaded))
    print(f"\nProcessing: {filename}")

    try:
        # Preprocess
        processed_img = preprocess_image(filename)
        # OCR
        raw_text = extract_text(processed_img)
        # Clean output
        cleaned = clean_text(raw_text)

        print("\n" + "="*50)
        print("EXTRACTED TEXT (after preprocessing & cleaning)")
        print("="*50)
        print(cleaned if cleaned else "[No text detected]")
    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    main()
```

Q1: Did the results improve?

For the handwritten/blurry image, the output still contains many artifacts, but the text is slightly more structured. The preprocessing helps with contrast, but handwritten text remains challenging for Tesseract.

**Is OCR enough to fully understand a document?**  
No, OCR alone is not enough.

**What OCR can do:**  
- Extract raw text from images (printed, clean, machine‑readable).  
- Provide bounding box coordinates for detected text.  
- Handle multiple languages if trained accordingly.

**What it cannot do:**  
- Understand the meaning or context of the text.  
- Distinguish between a date, a total amount, or a product name.  
- Interpret handwriting or damaged documents reliably.  
- Handle complex layouts (tables, forms, multi‑column) without additional logic.

**What additional AI might be needed:**  
- **Document Intelligence (Azure AI Document Intelligence)** – to understand document structure, extract key‑value pairs, and parse tables.  
- **Natural Language Processing (NLP)** – to interpret sentiment, extract entities, or classify content.  
- **Custom models** – to handle domain‑specific jargon or unique layouts.  
- **Human‑in‑the‑loop** – for validation when accuracy is critical (e.g., legal or medical documents).

```python
# Task 8 – Extended OCR with Highlighting, Number Extraction & Text Detection

# Install dependencies
!apt-get install -y tesseract-ocr
!pip install pytesseract pillow opencv-python matplotlib

import pytesseract
import cv2
import numpy as np
import re
import matplotlib.pyplot as plt
from google.colab import files
from PIL import Image

def upload_image():
    """Upload an image file."""
    print("Please upload an image:")
    uploaded = files.upload()
    return next(iter(uploaded))

def preprocess_image(image_path):
    """Basic preprocessing to improve OCR."""
    img = cv2.imread(image_path)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    return thresh, img

def highlight_text(image, data):
    """Draw bounding boxes around detected text."""
    img_copy = image.copy()
    for i in range(len(data['text'])):
        if int(data['conf'][i]) > 0:  # ignore low confidence
            (x, y, w, h) = (data['left'][i], data['top'][i],
                            data['width'][i], data['height'][i])
            cv2.rectangle(img_copy, (x, y), (x+w, y+h), (0, 255, 0), 2)
    return img_copy

def extract_numbers(text):
    """Extract numbers (integers and decimals) from text."""
    return re.findall(r'\d+\.?\d*', text)

def contains_text(text):
    """Return True if text contains at least one alphanumeric word."""
    return bool(re.search(r'[a-zA-Z0-9]', text))

def main():
    filename = upload_image()
    print(f"\nProcessing: {filename}")

    # Preprocess and get data
    processed_img, original_img = preprocess_image(filename)

    # Get OCR data with bounding boxes
    data = pytesseract.image_to_data(processed_img, output_type=pytesseract.Output.DICT)
    text = pytesseract.image_to_string(processed_img).strip()

    # 1. Highlight text areas
    highlighted = highlight_text(original_img, data)
    plt.imshow(cv2.cvtColor(highlighted, cv2.COLOR_BGR2RGB))
    plt.title("Highlighted Text Areas")
    plt.axis('off')
    plt.show()

    # 2. Extract numbers
    numbers = extract_numbers(text)
    print("\n--- Extracted Numbers ---")
    print(numbers if numbers else "[No numbers found]")

    # 3. Detect if image contains text
    has_text = contains_text(text)
    print(f"\nContains text? {has_text}")

    # Also show the raw extracted text
    print("\n--- Extracted Text ---")
    print(text if text else "[No text detected]")

if __name__ == "__main__":
    main()
```

Uses image_to_data to get coordinates and confidence.

Draws green rectangles around each detected text region.

Displays the annotated image using matplotlib.



2.Extract only numbers

Applies regex \d+\.?\d* to capture integers and decimal numbers from the extracted text.

Prints the list of numbers.

Detect if the image contains tex



3.Checks if the extracted text contains any alphanumeric character.

Outputs True or False.
